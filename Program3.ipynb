```python
#-A
import matplotlib.pyplot as plt
import numpy as np
x=np.array([0,10])
y=np.array([0,25])
#plot1
plt.subplot(2,3,1)
plt.plot(x,y,"o")
plt.xlabel("X-axis",color="Blue")
plt.ylabel("Y-axis",color="Blue")

plt.title("First Plot")
#plot2
x=np.array([1,6,1,6,1])
y=np.array([4,1,4,1,4])
plt.subplot(2,3,2)
plt.plot(x,marker="*",ms=20,mfc="blue",color="orange",lw="2.5")
plt.plot(y,ls="--")
#plot3
x=np.array([1,5,1,5])
plt.subplot(2,3,3)
plt.plot(x,marker="o",mfc="yellow",mec="red",ms=20,color="blue")
#plot4
x=np.array([1,2,3,4,5,6])
y=np.array([1,2,3,4,5,6])

plt.subplot(2,3,4)
plt.plot(x,y)
plt.title("Second plot")
plt.grid()
#plot5
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
plt.subplot(2,3,5)
plt.scatter(x, y)
x = np.array([2,2,8,1,15,8,12,9,7,3,11,4,7,14,12])
y = np.array([100,105,84,105,90,99,90,95,94,100,79,112,91,80,85])
plt.scatter(x, y,alpha=0.5)
#plot6
x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])
plt.subplot(2,3,6)
plt.bar(x, y, color = "red",width = 0.1)


plt.show()
```
