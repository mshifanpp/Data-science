```python
a=int(input("enter the first number"))
b=int(input("enter the second number"))
print("ARITHEMATIC OPERATION :")
print("Addition ",a+b)
print("Subtraction ",a-b)
print("Multiplication ",a*b)
if b==0:
    print("not divisible by zero")
else:
    print("Division ",a/b)
print("Exponation ",a**b)
```

```python
a=input("enter a string")
b=input("enter another string")
c=a+" "+b
print("STRING OPERATIONS ")
print("First string is ",a)
print("concatenation gives",c)
print("Repitation",a*3)
print("indexing",a[0])
print("silicing",a[1:3])
print("Length of a",len(a))
print("replace a to b",b.replace("a","b"))
print("String formatting: hello {}".format(a))
print("capitalize",a.capitalize())
print("Upper case :",a.upper())
print("Justify :",a.rjust(20))
print("To center :",a.center(20))
text=" python "
print("Text stripping :",text.strip())

```

```python
#lists
x=input("Enter the list of numbers: ").split( )
print(x)
print("value of position 1 :",x[1])
print("reversed list value",x[::-1])
print("list from 1 to 3",x[1:3])
print("last one",x[-1:])
x[2]='go'
print(x)
print("pop value",x.pop(1))
```

```python
#creat and print xs list
xs=[3,1,2]
print("the list=",xs)
print("Sliced number in possition 2=",xs[2])
print("last number in list=",xs[-1])
xs[2]='foo'
print(xs)
xs.append("5")
print("list after append",xs)
xs.pop(2)
print("list after poped",xs)
```

```python
#slicing
x=[0,1,2,3,4,5]
print(x[1:5])
print(x[:6])
print(x[2:])
print(x[1:5:2])
```
